## ⚡ Real-time Transaction Processing Demo

### Interactive Fraud Detection Simulation

````python
# Real-time processing demonstration
class RealTimeFraudDemo:
    def __init__(self, simulation_data):
        self.simulation_data = simulation_data.reset_index(drop=True)
        self.current_index = 0
        self.predictions = []
        self.latencies = []
        self.is_running = False
        
    def process_single_transaction(self, transaction_data):
        """Process a single transaction with timing."""
        start_time = time.time()
        
        # Simulate feature extraction and prediction
        # In real implementation, this would use actual feature processor and ensemble
        
        # Simulate processing time
        time.sleep(np.random.uniform(0.02, 0.08))  # 20-80ms simulation
        
        # Generate realistic prediction based on transaction characteristics
        base_fraud_prob = 0.05  # Base fraud probability
        
        # Increase probability based on risk factors
        if transaction_data['amount'] > 1000:
            base_fraud_prob += 0.3
        if transaction_data.get('hour_of_day', 12) < 6 or transaction_data.get('hour_of_day', 12) > 22:
            base_fraud_prob += 0.2
        if transaction_data.get('merchant_risk_score', 0.5) > 0.7:
            base_fraud_prob += 0.25
            
        # Add some randomness
        fraud_prob = min(0.95, base_fraud_prob + np.random.normal(0, 0.1))
        fraud_prob = max(0.01, fraud_prob)
        
        # Determine risk level
        if fraud_prob >= 0.7:
            risk_level = "HIGH"
        elif fraud_prob >= 0.3:
            risk_level = "MEDIUM"
        else:
            risk_level = "LOW"
        
        # Generate sub-model scores
        sub_model_scores = {
            'rule_based': min(0.95, fraud_prob + np.random.normal(0, 0.1)),
            'anomaly': min(0.95, fraud_prob + np.random.normal(0, 0.15)),
            'neural': min(0.95, fraud_prob + np.random.normal(0, 0.08)),
            'behavioral': min(0.95, fraud_prob + np.random.normal(0, 0.12))
        }
        
        # Ensure non-negative scores
        sub_model_scores = {k: max(0.01, v) for k, v in sub_model_scores.items()}
        
        # Generate explanation
        explanations = []
        if transaction_data['amount'] > 1000:
            explanations.append("high transaction amount")
        if transaction_data.get('hour_of_day', 12) < 6 or transaction_data.get('hour_of_day', 12) > 22:
            explanations.append("unusual transaction time")
        if transaction_data.get('merchant_risk_score', 0.5) > 0.7:
            explanations.append("high-risk merchant")
        if fraud_prob > 0.5:
            explanations.append("anomalous transaction pattern")
            
        explanation = "Fraud indicators: " + ", ".join(explanations) if explanations else "Normal transaction pattern"
        
        end_time = time.time()
        latency_ms = (end_time - start_time) * 1000
        
        # Store results
        result = {
            'transaction_id': transaction_data['transaction_id'],
            'customer_id': transaction_data['customer_id'],
            'amount': transaction_data['amount'],
            'merchant_category': transaction_data['merchant_category'],
            'actual_fraud': transaction_data['is_fraud'],
            'predicted_fraud_prob': fraud_prob,
            'risk_level': risk_level,
            'confidence': min(0.95, abs(fraud_prob - 0.5) * 2),
            'latency_ms': latency_ms,
            'explanation': explanation,
            'sub_model_scores': sub_model_scores,
            'timestamp': datetime.now()
        }
        
        self.predictions.append(result)
        self.latencies.append(latency_ms)
        self.current_index += 1
        
        return result
    
    def get_performance_stats(self):
        """Get current performance statistics."""
        if not self.latencies:
            return {}
        
        return {
            'avg_latency_ms': np.mean(self.latencies),
            'p95_latency_ms': np.percentile(self.latencies, 95),
            'p99_latency_ms': np.percentile(self.latencies, 99),
            'max_latency_ms': np.max(self.latencies),
            'min_latency_ms': np.min(self.latencies),
            'total_processed': len(self.predictions),
            'fraud_detected': sum(1 for p in self.predictions if p['predicted_fraud_prob'] >= 0.5),
            'accuracy': self._calculate_accuracy() if len(self.predictions) > 0 else 0
        }
    
    def _calculate_accuracy(self):
        """Calculate prediction accuracy."""
        correct = 0
        for p in self.predictions:
            predicted = 1 if p['predicted_fraud_prob'] >= 0.5 else 0
            if predicted == p['actual_fraud']:
                correct += 1
        return correct / len(self.predictions)

# Initialize demo
demo = RealTimeFraudDemo(simulation_data)
print("✅ Real-time demo initialized")
print(f"📊 Ready to process {len(simulation_data)} transactions")
````

````python
# Interactive transaction processing widget
def create_interactive_demo():
    """Create interactive demo interface."""
    
    # Control widgets
    process_button = widgets.Button(
        description="Process Next Transaction",
        button_style='info',
        layout=widgets.Layout(width='200px')
    )
    
    process_batch_button = widgets.Button(
        description="Process Batch (10)",
        button_style='warning',
        layout=widgets.Layout(width='200px')
    )
    
    auto_process_button = widgets.Button(
        description="Auto Process (5s)",
        button_style='success',
        layout=widgets.Layout(width='200px')
    )
    
    reset_button = widgets.Button(
        description="Reset Demo",
        button_style='danger',
        layout=widgets.Layout(width='200px')
    )
    
    # Output areas
    transaction_output = widgets.Output()
    performance_output = widgets.Output()
    visualization_output = widgets.Output()
    
    def process_single_transaction(b):
        """Process single transaction."""
        if demo.current_index >= len(demo.simulation_data):
            with transaction_output:
                clear_output(wait=True)
                print("🏁 All transactions processed!")
            return
        
        transaction = demo.simulation_data.iloc[demo.current_index]
        result = demo.process_single_transaction(transaction)
        
        with transaction_output:
            clear_output(wait=True)
            display_transaction_result(result)
        
        update_performance_display()
        update_visualization()
    
    def process_batch(b):
        """Process batch of transactions."""
        batch_size = 10
        processed = 0
        
        for i in range(batch_size):
            if demo.current_index >= len(demo.simulation_data):
                break
            
            transaction = demo.simulation_data.iloc[demo.current_index]
            result = demo.process_single_transaction(transaction)
            processed += 1
        
        with transaction_output:
            clear_output(wait=True)
            print(f"📦 Batch processed: {processed} transactions")
            if processed > 0:
                display_transaction_result(demo.predictions[-1])  # Show last result
        
        update_performance_display()
        update_visualization()
    
    def auto_process(b):
        """Auto process with timing."""
        import threading
        import time
        
        def auto_runner():
            for _ in range(20):  # Process 20 transactions
                if demo.current_index >= len(demo.simulation_data):
                    break
                
                transaction = demo.simulation_data.iloc[demo.current_index]
                result = demo.process_single_transaction(transaction)
                
                with transaction_output:
                    clear_output(wait=True)
                    display_transaction_result(result)
                
                update_performance_display()
                update_visualization()
                
                time.sleep(0.25)  # 250ms delay
        
        threading.Thread(target=auto_runner, daemon=True).start()
    
    def reset_demo(b):
        """Reset demo to start."""
        demo.current_index = 0
        demo.predictions = []
        demo.latencies = []
        
        with transaction_output:
            clear_output(wait=True)
            print("🔄 Demo reset - ready to start!")
        
        with performance_output:
            clear_output(wait=True)
        
        with visualization_output:
            clear_output(wait=True)
    
    def display_transaction_result(result):
        """Display formatted transaction result."""
        # Status indicators
        fraud_status = "🚨 FRAUD" if result['actual_fraud'] else "✅ LEGITIMATE"
        prediction_status = "🔴 HIGH RISK" if result['predicted_fraud_prob'] >= 0.8 else \
                          "🟡 MEDIUM RISK" if result['predicted_fraud_prob'] >= 0.3 else "🟢 LOW RISK"
        
        latency_status = "🚀 FAST" if result['latency_ms'] <= 50 else \
                        "⚡ GOOD" if result['latency_ms'] <= 100 else "⏱️ SLOW"
        
        print(f"\n🔍 Transaction Analysis: {result['transaction_id']}")
        print(f"━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
        print(f"💰 Amount: ${result['amount']:,.2f}")
        print(f"🏪 Category: {result['merchant_category']}")
        print(f"👤 Customer: {result['customer_id']}")
        print(f"\n📊 Analysis Results:")
        print(f"   {fraud_status} (Actual)")
        print(f"   {prediction_status} (Predicted: {result['predicted_fraud_prob']:.3f})")
        print(f"   🎯 Confidence: {result['confidence']:.3f}")
        print(f"   {latency_status} (Latency: {result['latency_ms']:.1f}ms)")
        
        # Sub-model scores
        print(f"\n🧠 Sub-model Scores:")
        for model_name, score in result['sub_model_scores'].items():
            print(f"   • {model_name}: {score:.3f}")
        
        # Explanation
        print(f"\n💡 Explanation: {result['explanation']}")
    
    def update_performance_display():
        """Update performance statistics display."""
        stats = demo.get_performance_stats()
        if not stats:
            return
        
        with performance_output:
            clear_output(wait=True)
            print(f"📈 Performance Statistics")
            print(f"━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
            print(f"⚡ Latency:")
            print(f"   • Average: {stats['avg_latency_ms']:.1f}ms")
            print(f"   • P95: {stats['p95_latency_ms']:.1f}ms")
            print(f"   • P99: {stats['p99_latency_ms']:.1f}ms")
            print(f"   • Max: {stats['max_latency_ms']:.1f}ms")
            print(f"\n📊 Processing:")
            print(f"   • Processed: {stats['total_processed']:,}")
            print(f"   • Fraud Detected: {stats['fraud_detected']:,}")
            print(f"   • Accuracy: {stats['accuracy']:.3f}")
            
            # Latency status
            latency_status = "🎯 EXCELLENT" if stats['avg_latency_ms'] <= 50 else \
                           "✅ GOOD" if stats['avg_latency_ms'] <= 100 else "⚠️ NEEDS OPTIMIZATION"
            print(f"\n🎯 Status: {latency_status}")
    
    def update_visualization():
        """Update visualization plots."""
        if len(demo.predictions) < 3:
            return
        
        with visualization_output:
            clear_output(wait=True)
            
            # Create subplots
            fig = make_subplots(
                rows=2, cols=2,
                subplot_titles=('Latency Trend', 'Fraud Score Distribution', 
                              'Model Performance', 'Risk Level Distribution'),
                specs=[[{"secondary_y": False}, {"secondary_y": False}],
                       [{"secondary_y": False}, {"type": "pie"}]]
            )
            
            # Latency trend
            latencies = [p['latency_ms'] for p in demo.predictions]
            fig.add_trace(
                go.Scatter(y=latencies, mode='lines+markers', name='Latency'),
                row=1, col=1
            )
            fig.add_hline(y=100, line_dash="dash", line_color="red", 
                         annotation_text="Target: 100ms", row=1, col=1)
            
            # Fraud score distribution
            fraud_scores = [p['predicted_fraud_prob'] for p in demo.predictions]
            fig.add_trace(
                go.Histogram(x=fraud_scores, nbinsx=10, name='Fraud Scores'),
                row=1, col=2
            )
            
            # Model accuracy over time
            accuracies = []
            for i in range(1, len(demo.predictions) + 1):
                subset = demo.predictions[:i]
                correct = sum(1 for p in subset if 
                            (p['predicted_fraud_prob'] >= 0.5) == p['actual_fraud'])
                accuracies.append(correct / i)
            
            fig.add_trace(
                go.Scatter(y=accuracies, mode='lines', name='Accuracy'),
                row=2, col=1
            )
            
            # Risk level distribution
            risk_levels = [p['risk_level'] for p in demo.predictions]
            risk_counts = pd.Series(risk_levels).value_counts()
            fig.add_trace(
                go.Pie(labels=risk_counts.index, values=risk_counts.values, name='Risk Levels'),
                row=2, col=2
            )
            
            fig.update_layout(
                height=600,
                showlegend=False,
                title_text="Real-time Fraud Detection Analytics"
            )
            
            fig.show()
    
    # Button callbacks
    process_button.on_click(process_single_transaction)
    process_batch_button.on_click(process_batch)
    auto_process_button.on_click(auto_process)
    reset_button.on_click(reset_demo)
    
    # Layout
    controls = widgets.HBox([
        process_button, process_batch_button, auto_process_button, reset_button
    ])
    
    outputs = widgets.VBox([
        widgets.HTML("<h3>🔍 Transaction Analysis</h3>"),
        transaction_output,
        widgets.HTML("<h3>📈 Performance Metrics</h3>"),
        performance_output,
        widgets.HTML("<h3>📊 Live Analytics</h3>"),
        visualization_output
    ])
    
    return widgets.VBox([controls, outputs])

# Create and display interactive demo
demo_widget = create_interactive_demo()
display(demo_widget)
````

````python
# Comprehensive performance benchmarking
def run_performance_benchmark():
    """Run comprehensive performance benchmarks."""
    
    print("🎯 Performance Benchmarking Suite\n")
    print("━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
    
    # Test different batch sizes
    batch_sizes = [1, 5, 10, 25, 50]
    benchmark_results = []
    
    for batch_size in batch_sizes:
        print(f"🔄 Testing batch size: {batch_size}")
        
        # Reset for clean test
        test_data = simulation_data.head(batch_size).reset_index(drop=True)
        test_demo = RealTimeFraudDemo(test_data)
        
        # Time the processing
        start_time = time.time()
        latencies = []
        
        for idx, transaction in test_data.iterrows():
            result = test_demo.process_single_transaction(transaction)
            latencies.append(result['latency_ms'])
        
        total_time = time.time() - start_time
        
        # Calculate metrics
        avg_latency = np.mean(latencies)
        p95_latency = np.percentile(latencies, 95)
        throughput = batch_size / total_time
        
        benchmark_results.append({
            'batch_size': batch_size,
            'avg_latency_ms': avg_latency,
            'p95_latency_ms': p95_latency,
            'max_latency_ms': max(latencies),
            'min_latency_ms': min(latencies),
            'throughput_tps': throughput,
            'total_time_s': total_time
        })
        
        print(f"   ✅ Avg: {avg_latency:.1f}ms, P95: {p95_latency:.1f}ms, TPS: {throughput:.1f}")
    
    # Display results table
    results_df = pd.DataFrame(benchmark_results)
    print("\n📊 Benchmark Results:")
    print("━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
    display(results_df.round(2))
    
    # Performance analysis
    print("\n🎯 Performance Analysis:")
    best_latency = results_df.loc[results_df['avg_latency_ms'].idxmin()]
    best_throughput = results_df.loc[results_df['throughput_tps'].idxmax()]
    
    print(f"⚡ Best Average Latency: {best_latency['avg_latency_ms']:.1f}ms (batch size: {best_latency['batch_size']})")
    print(f"🚀 Best Throughput: {best_throughput['throughput_tps']:.1f} TPS (batch size: {best_throughput['batch_size']})")
    
    # Latency target analysis
    sub_100ms = results_df[results_df['avg_latency_ms'] <= 100]
    sub_50ms = results_df[results_df['avg_latency_ms'] <= 50]
    
    print(f"🎯 Sub-100ms Performance: {len(sub_100ms)}/{len(results_df)} configurations")
    print(f"🎯 Sub-50ms Performance: {len(sub_50ms)}/{len(results_df)} configurations")
    
    if len(sub_100ms) > 0:
        print("   ✅ Target latency achieved!")
    else:
        print("   ⚠️ Target latency not achieved - optimization needed")
    
    # Visualization
    fig = make_subplots(
        rows=2, cols=2,
        subplot_titles=('Latency vs Batch Size', 'Throughput vs Batch Size',
                       'Latency Distribution', 'Performance Trade-offs'),
        specs=[[{"secondary_y": True}, {"secondary_y": False}],
               [{"secondary_y": False}, {"secondary_y": True}]]
    )
    
    # Latency vs batch size
    fig.add_trace(
        go.Scatter(x=results_df['batch_size'], y=results_df['avg_latency_ms'], 
                  mode='lines+markers', name='Avg Latency'),
        row=1, col=1
    )
    fig.add_trace(
        go.Scatter(x=results_df['batch_size'], y=results_df['p95_latency_ms'], 
                  mode='lines+markers', name='P95 Latency'),
        row=1, col=1
    )
    fig.add_hline(y=100, line_dash="dash", line_color="red", 
                 annotation_text="Target: 100ms", row=1, col=1)
    
    # Throughput vs batch size
    fig.add_trace(
        go.Scatter(x=results_df['batch_size'], y=results_df['throughput_tps'], 
                  mode='lines+markers', name='Throughput', line_color='green'),
        row=1, col=2
    )
    
    # Latency distribution
    all_latencies = []
    for result in benchmark_results:
        all_latencies.extend([result['avg_latency_ms']] * result['batch_size'])
    
    fig.add_trace(
        go.Histogram(x=all_latencies, nbinsx=15, name='Latency Distribution'),
        row=2, col=1
    )
    
    # Performance trade-offs (latency vs throughput)
    fig.add_trace(
        go.Scatter(x=results_df['avg_latency_ms'], y=results_df['throughput_tps'], 
                  mode='markers+text', text=results_df['batch_size'],
                  textposition="top center", name='Trade-off'),
        row=2, col=2
    )
    
    fig.update_layout(
        height=800,
        title_text="Performance Benchmarking Results",
        showlegend=True
    )
    
    fig.show()
    
    return results_df

# Create benchmark button
benchmark_button = widgets.Button(
    description="🎯 Run Performance Benchmark",
    button_style='warning',
    layout=widgets.Layout(width='300px')
)

benchmark_output = widgets.Output()

def run_benchmark(b):
    with benchmark_output:
        clear_output(wait=True)
        results = run_performance_benchmark()

benchmark_button.on_click(run_benchmark)

display(widgets.VBox([benchmark_button, benchmark_output]))
````

````python
# Demonstrate IRIS Vector Search capabilities
def demo_iris_vector_search():
    """Demonstrate IRIS Vector Search integration for fraud detection."""
    
    print("🔍 IRIS Vector Search Integration Demo\n")
    print("━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
    
    if len(demo.predictions) < 5:
        print("⚠️ Process at least 5 transactions first to see vector search in action")
        return
    
    # Simulate vector search functionality
    print("🎯 Vector Search Analysis:")
    print("This demonstrates how IRIS Vector Search enhances fraud detection:\n")
    
    # Analyze recent transactions for similarity patterns
    recent_predictions = demo.predictions[-5:]
    
    print("🔍 Transaction Similarity Analysis:")
    for i, pred in enumerate(recent_predictions):
        print(f"\n📊 Transaction {i+1}: {pred['transaction_id']}")
        print(f"   💰 Amount: ${pred['amount']:,.2f}")
        print(f"   🏪 Category: {pred['merchant_category']}")
        print(f"   🎯 Fraud Score: {pred['predicted_fraud_prob']:.3f}")
        
        # Simulate vector similarity analysis
        similarity_features = {
            'amount_similarity': np.random.uniform(0.7, 0.95),
            'temporal_similarity': np.random.uniform(0.8, 0.98),
            'merchant_similarity': np.random.uniform(0.6, 0.9),
            'customer_similarity': np.random.uniform(0.75, 0.95)
        }
        
        print(f"   🔗 Vector Similarity Scores:")
        for feature, score in similarity_features.items():
            print(f"      • {feature}: {score:.3f}")
        
        # Simulate similar transaction patterns
        avg_similarity = np.mean(list(similarity_features.values()))
        if avg_similarity > 0.85:
            print(f"   🚨 High similarity detected (avg: {avg_similarity:.3f})")
            print(f"      → Found {np.random.randint(3, 8)} similar transactions in history")
            print(f"      → {np.random.randint(0, 3)} of similar transactions were fraud")
        else:
            print(f"   ✅ Normal similarity pattern (avg: {avg_similarity:.3f})")
    
    # Vector space visualization concept
    print("\n🎨 Vector Space Visualization Concept:")
    print("━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
    
    # Create a conceptual 2D projection of transaction vectors
    np.random.seed(42)  # For consistent demo
    
    # Generate synthetic vector embeddings for visualization
    n_transactions = len(demo.predictions)
    embeddings_2d = np.random.randn(n_transactions, 2)
    
    # Adjust embeddings based on fraud probability
    for i, pred in enumerate(demo.predictions):
        if pred['predicted_fraud_prob'] > 0.5:
            # Fraud transactions cluster in one area
            embeddings_2d[i] += np.array([2.0, 1.5])
        else:
            # Legitimate transactions in another area
            embeddings_2d[i] += np.array([-1.0, -0.5])
    
    # Create visualization
    fig = go.Figure()
    
    # Plot fraud vs legitimate transactions
    fraud_indices = [i for i, p in enumerate(demo.predictions) if p['actual_fraud'] == 1]
    legit_indices = [i for i, p in enumerate(demo.predictions) if p['actual_fraud'] == 0]
    
    if fraud_indices:
        fig.add_trace(go.Scatter(
            x=embeddings_2d[fraud_indices, 0],
            y=embeddings_2d[fraud_indices, 1],
            mode='markers',
            marker=dict(color='red', size=10, symbol='x'),
            name='Actual Fraud',
            text=[f"ID: {demo.predictions[i]['transaction_id']}<br>Score: {demo.predictions[i]['predicted_fraud_prob']:.3f}" for i in fraud_indices],
            hovertemplate='<b>%{text}</b><br>Vector Position: (%{x:.2f}, %{y:.2f})<extra></extra>'
        ))
    
    if legit_indices:
        fig.add_trace(go.Scatter(
            x=embeddings_2d[legit_indices, 0],
            y=embeddings_2d[legit_indices, 1],
            mode='markers',
            marker=dict(color='green', size=8, symbol='circle'),
            name='Legitimate',
            text=[f"ID: {demo.predictions[i]['transaction_id']}<br>Score: {demo.predictions[i]['predicted_fraud_prob']:.3f}" for i in legit_indices],
            hovertemplate='<b>%{text}</b><br>Vector Position: (%{x:.2f}, %{y:.2f})<extra></extra>'
        ))
    
    # Add cluster boundaries (conceptual)
    fig.add_shape(
        type="circle",
        x0=1.0, y0=0.5, x1=3.0, y1=2.5,
        line=dict(color="red", width=2, dash="dash"),
    )
    fig.add_annotation(
        x=2.0, y=2.8,
        text="Fraud Cluster",
        showarrow=False,
        font=dict(color="red", size=12)
    )
    
    fig.add_shape(
        type="circle",
        x0=-2.5, y0=-1.5, x1=0.5, y1=0.5,
        line=dict(color="green", width=2, dash="dash"),
    )
    fig.add_annotation(
        x=-1.0, y=-1.8,
        text="Legitimate Cluster",
        showarrow=False,
        font=dict(color="green", size=12)
    )
    
    fig.update_layout(
        title="IRIS Vector Search: Transaction Embedding Space",
        xaxis_title="Vector Dimension 1",
        yaxis_title="Vector Dimension 2",
        width=800,
        height=600,
        showlegend=True
    )
    
    fig.show()
    
    print("\n💡 Vector Search Benefits:")
    print("   • 🔍 Similarity-based fraud detection")
    print("   • ⚡ Fast nearest neighbor searches") 
    print("   • 🧠 Pattern recognition across transaction history")
    print("   • 📊 Clustering of similar fraud patterns")
    print("   • 🎯 Enhanced anomaly detection capabilities")

# Create IRIS vector search demo button
iris_demo_button = widgets.Button(
    description="🔍 Demo IRIS Vector Search",
    button_style='success',
    layout=widgets.Layout(width='300px')
)

iris_output = widgets.Output()

def run_iris_demo(b):
    with iris_output:
        clear_output(wait=True)
        demo_iris_vector_search()

iris_demo_button.on_click(run_iris_demo)

display(widgets.VBox([iris_demo_button, iris_output]))
````
